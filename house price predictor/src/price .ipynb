### Bagging and Boosting Techniques

`Gradient Boosting Regressor` and `XGBoost Regressor` are both powerful boosting techniques that build an ensemble of weak learners sequentially, with each new learner correcting errors made by the previous ones.

**Bagging (Bootstrap Aggregating)**, on the other hand, involves training multiple models (usually of the same type) independently on different bootstrap samples of the training data, and then combining their predictions (e.g., by averaging for regression). A common example is the `BaggingRegressor` from scikit-learn.

While Gradient Boosting and XGBoost are highly effective, explicitly trying a `BaggingRegressor` with a strong base estimator (like a `DecisionTreeRegressor`) can sometimes provide different insights or performance, especially if the individual models are diverse.

Let's evaluate a `BaggingRegressor`.

```python
from sklearn.ensemble import BaggingRegressor

# Add Bagging Regressor to the models dictionary
# Using a Decision Tree Regressor as the base estimator
models["Bagging Regressor (Decision Tree)"] = BaggingRegressor(
    estimator=DecisionTreeRegressor(random_state=42),
    n_estimators=100, # Number of base estimators
    random_state=42,
    n_jobs=-1 # Use all available cores
)

print("--- Evaluating Models with StandardScaler (including Bagging) ---")
# Re-evaluating all models to ensure 'results' dictionary is updated with Bagging Regressor
for name, model in models.items():
    print(f"\n--- Evaluating {name} ---")
    # Create a pipeline with StandardScaler and the current model
    pipeline = Pipeline([
        ('scaler', StandardScaler()),
        ('model', model)
    ])

    # Fit the pipeline
    pipeline.fit(X_train, y_train)

    # Make predictions
    y_pred = pipeline.predict(X_test)

    # Evaluate the model
    r2 = r2_score(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)

    print(f"R-squared: {r2}")
    print(f"Mean Squared Error: {mse}")

    results[name] = {"R-squared": r2, "Mean Squared Error": mse}

# Find and print the best model based on R-squared
best_model_name = max(results, key=lambda k: results[k]['R-squared'])
print(f"\n--- Best Model based on R-squared: {best_model_name} ---")
print(f"R-squared: {results[best_model_name]['R-squared']}")
print(f"Mean Squared Error: {results[best_model_name]['Mean Squared Error']}")
```

### Applying Stacking Ensemble with Gradient Boosting Regressor as Meta-Model

Stacking is a powerful ensemble technique where the predictions of several base estimators (level-0 models) are used as input features for a meta-estimator (level-1 model). We'll use `LinearRegression`, `RandomForestRegressor`, and `XGBRegressor` as our base models, and a `GradientBoostingRegressor` as the final meta-model to make the ultimate prediction.

```python
from sklearn.ensemble import StackingRegressor, GradientBoostingRegressor, RandomForestRegressor
from sklearn.linear_model import LinearRegression
from xgboost import XGBRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import r2_score, mean_squared_error

# Define base estimators (level-0 models), each with its own StandardScaler
# This ensures that each base model receives scaled data independently
estimators = [
    ('lr', Pipeline([
        ('scaler', StandardScaler()),
        ('model', LinearRegression())
    ])),
    ('rf', Pipeline([
        ('scaler', StandardScaler()),
        ('model', RandomForestRegressor(random_state=42))
    ])),
    ('xgb', Pipeline([
        ('scaler', StandardScaler()),
        ('model', XGBRegressor(random_state=42))
    ]))
]

# Define the meta-estimator (level-1 model)
# The GradientBoostingRegressor will learn to combine the predictions of the base estimators
final_estimator = GradientBoostingRegressor(random_state=42)

# Create the StackingRegressor pipeline
# This pipeline will handle the training of base estimators and the meta-estimator
stacked_model = StackingRegressor(
    estimators=estimators,
    final_estimator=final_estimator,
    cv=5, # Number of cross-validation folds to train base models
    n_jobs=-1 # Use all available cores
)

# Add the stacked model to our dictionary for evaluation
models["Stacking Regressor (GBM Meta)"] = stacked_model

print("--- Evaluating Models (including Stacking Regressor) ---")
# Re-evaluating all models to ensure 'results' dictionary is updated with the new stacked model
for name, model in models.items():
    print(f"\n--- Evaluating {name} ---")
    # For stacking, the internal pipelines for base estimators already handle scaling.
    # So, we'll create a simple pipeline here that just contains the model if it's already a pipeline (like StackingRegressor).
    # For other models, we'll keep the StandardScaler pipeline as before.
    if isinstance(model, StackingRegressor):
        pipeline = model # The stacking regressor itself is a complex pipeline
    else:
        pipeline = Pipeline([
            ('scaler', StandardScaler()),
            ('model', model)
        ])

    # Fit the pipeline
    pipeline.fit(X_train, y_train)

    # Make predictions
    y_pred = pipeline.predict(X_test)

    # Evaluate the model
    r2 = r2_score(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)

    print(f"R-squared: {r2}")
    print(f"Mean Squared Error: {mse}")

    results[name] = {"R-squared": r2, "Mean Squared Error": mse}

# Find and print the best model based on R-squared
best_model_name = max(results, key=lambda k: results[k]['R-squared'])
print(f"\n--- Best Model based on R-squared: {best_model_name} ---")
print(f"R-squared: {results[best_model_name]['R-squared']}")
print(f"Mean Squared Error: {results[best_model_name]['Mean Squared Error']}")
```

### Advanced Hyperparameter Tuning with Optuna for GradientBoostingRegressor

To further optimize the `GradientBoostingRegressor`, we'll use Optuna, an automatic hyperparameter optimization framework. Optuna intelligently searches for the best combination of hyperparameters to minimize (or maximize) a given objective function. In our case, the objective will be to minimize the Mean Squared Error (MSE) on the validation set.

```python
# Install Optuna if not already installed
!pip install optuna --quiet
```

```python
import optuna
from sklearn.model_selection import KFold
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

def objective(trial):
    # Define the search space for GradientBoostingRegressor hyperparameters
    n_estimators = trial.suggest_int('n_estimators', 100, 1000, step=100)
    learning_rate = trial.suggest_float('learning_rate', 0.01, 0.2, log=True)
    max_depth = trial.suggest_int('max_depth', 3, 10)
    min_samples_split = trial.suggest_int('min_samples_split', 2, 20)
    min_samples_leaf = trial.suggest_int('min_samples_leaf', 1, 10)
    subsample = trial.suggest_float('subsample', 0.6, 1.0, step=0.1)

    # Create a GradientBoostingRegressor with the suggested hyperparameters
    gbm = GradientBoostingRegressor(
        n_estimators=n_estimators,
        learning_rate=learning_rate,
        max_depth=max_depth,
        min_samples_split=min_samples_split,
        min_samples_leaf=min_samples_leaf,
        subsample=subsample,
        random_state=42
    )

    # Create a pipeline with StandardScaler and the GradientBoostingRegressor
    pipeline = Pipeline([
        ('scaler', StandardScaler()),
        ('model', gbm)
    ])

    # Use K-Fold cross-validation to evaluate the model
    kf = KFold(n_splits=5, shuffle=True, random_state=42)
    mse_scores = []

    for train_index, val_index in kf.split(X_train):
        X_train_fold, X_val_fold = X_train.iloc[train_index], X_train.iloc[val_index]
        y_train_fold, y_val_fold = y_train.iloc[train_index], y_train.iloc[val_index]

        pipeline.fit(X_train_fold, y_train_fold)
        y_pred_val = pipeline.predict(X_val_fold)
        mse_scores.append(mean_squared_error(y_val_fold, y_pred_val))

    # Return the average MSE across all folds
    return np.mean(mse_scores)

# Create an Optuna study and optimize the objective function
print("Starting Optuna hyperparameter optimization...")
study = optuna.create_study(direction='minimize') # We want to minimize MSE
study.optimize(objective, n_trials=50) # Run 50 trials

print("\n--- Optuna Optimization Results ---")
print(f"Best trial: {study.best_trial.value:.4f} (MSE)")
print("Best hyperparameters found:")
for key, value in study.best_params.items():
    print(f"  {key}: {value}")

# Train the final GradientBoostingRegressor with the best hyperparameters from Optuna
best_gbm_optuna = GradientBoostingRegressor(
    **study.best_params,
    random_state=42
)

# Create a pipeline for the optimized model
optimized_gbm_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('model', best_gbm_optuna)
])

# Fit the pipeline on the full training data
optimized_gbm_pipeline.fit(X_train, y_train)

# Make predictions on the test set
y_pred_optuna_gbm = optimized_gbm_pipeline.predict(X_test)

# Evaluate the optimized model
r2_optuna_gbm = r2_score(y_test, y_pred_optuna_gbm)
mse_optuna_gbm = mean_squared_error(y_test, y_pred_optuna_gbm)

print(f"\n--- Optimized GradientBoostingRegressor (Optuna) Performance on Test Set ---")
print(f"R-squared: {r2_optuna_gbm}")
print(f"Mean Squared Error: {mse_optuna_gbm}")

# Add to results for comparison
results["Gradient Boosting Regressor (Optuna Tuned)"] = {"R-squared": r2_optuna_gbm, "Mean Squared Error": mse_optuna_gbm}

# Update and print the best model based on R-squared
best_model_name = max(results, key=lambda k: results[k]['R-squared'])
print(f"\n--- Best Model Overall based on R-squared: {best_model_name} ---")
print(f"R-squared: {results[best_model_name]['R-squared']}")
print(f"Mean Squared Error: {results[best_model_name]['Mean Squared Error']}")
```

### Creating a Gradio App to Interact with the Model

Now that we have our `model.pkl` file, let's create a simple web application using Gradio. This app will allow us to input various house features and get a price prediction from our deployed model.

```python
import gradio as gr
import joblib
import pandas as pd
import numpy as np

# Load the saved model pipeline
loaded_pipeline = joblib.load('model.pkl')

# Define the feature columns as they were used during model training
# This list must match the order and names of features X_train was composed of.
FEATURE_COLUMNS = [
    'area', 'bedrooms', 'bathrooms', 'stories', 'mainroad', 'guestroom',
    'basement', 'hotwaterheating', 'airconditioning', 'parking',
    'prefarea', 'furnishingstatus'
]

# Define the prediction function that will be used by Gradio
def predict_price(
    area,
    bedrooms,
    bathrooms,
    stories,
    mainroad,
    guestroom,
    basement,
    hotwaterheating,
    airconditioning,
    parking,
    prefarea,
    furnishingstatus
):
    # Map 'Yes'/'No' inputs to 1/0 for binary categorical features
    mainroad_val = 1 if mainroad == 'Yes' else 0
    guestroom_val = 1 if guestroom == 'Yes' else 0
    basement_val = 1 if basement == 'Yes' else 0
    hotwaterheating_val = 1 if hotwaterheating == 'Yes' else 0
    airconditioning_val = 1 if airconditioning == 'Yes' else 0
    prefarea_val = 1 if prefarea == 'Yes' else 0

    # Map furnishingstatus to its encoded numerical value
    furnishing_map = {'furnished': 0, 'semi-furnished': 1, 'unfurnished': 2}
    furnishingstatus_val = furnishing_map.get(furnishingstatus, 2) # Default to 'unfurnished' if not found

    # Create a DataFrame from the inputs, ensuring correct column order
    input_data = pd.DataFrame([[area, bedrooms, bathrooms, stories, mainroad_val, guestroom_val,
                                  basement_val, hotwaterheating_val, airconditioning_val, parking,
                                  prefarea_val, furnishingstatus_val]],
                              columns=FEATURE_COLUMNS) # Use the defined FEATURE_COLUMNS

    # Make prediction using the loaded pipeline
    prediction = loaded_pipeline.predict(input_data)[0]

    return f"Predicted Price: {prediction:,.2f} INR"

# Define Gradio input components
inputs = [
    gr.Number(label="Area (sq ft)", value=5900),
    gr.Number(label="Bedrooms", value=4, precision=0),
    gr.Number(label="Bathrooms", value=2, precision=0),
    gr.Number(label="Stories", value=2, precision=0),
    gr.Radio(choices=["No", "Yes"], label="Main Road Access", value="No"),
    gr.Radio(choices=["No", "Yes"], label="Guest Room", value="No"),
    gr.Radio(choices=["No", "Yes"], label="Basement", value="Yes"),
    gr.Radio(choices=["No", "Yes"], label="Hot Water Heating", value="No"),
    gr.Radio(choices=["No", "Yes"], label="Air Conditioning", value="No"),
    gr.Number(label="Parking Spaces", value=1, precision=0),
    gr.Radio(choices=["No", "Yes"], label="Preferred Area", value="No"),
    gr.Dropdown(choices=["furnished", "semi-furnished", "unfurnished"], label="Furnishing Status", value="semi-furnished")
]

# Create the Gradio interface
interface = gr.Interface(
    fn=predict_price,
    inputs=inputs,
    outputs="text",
    title="House Price Prediction",
    description="Enter the features of the house to get a price prediction."
)

# Launch the app
if __name__ == "__main__":
    interface.launch(share=True)
```
