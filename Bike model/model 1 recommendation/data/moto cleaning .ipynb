```python
def clean_transmission_type(df):
    df['transmission_type'] = df['transmission_type'].str.lower()

    def extract_transmission_keyword(text):
        s_text = str(text)
        parts = s_text.replace(' ', '-').split('-')
        if len(parts) == 1:
            return parts[0]
        elif len(parts) == 2:
            return parts[1]
        else:
            return parts[-1]

    df['transmission_type'] = df['transmission_type'].apply(extract_transmission_keyword)
    return df
```

```python
df = clean_transmission_type(df)
df['transmission_type'].unique()
```

```python
def clean_price(df):
    df['price'] = df['price'].str.replace('₹', '', regex=False).str.replace('$','',regex=False).str.replace('Rs.','',regex=False).str.replace(',','')
    df['price']
```

```python
# 1. Standardize initial cleaning (remove currency symbols, Rs., commas, INR, and specific parenthetical text)
df['price'] = df['price'].astype(str).str.lower() \
                        .str.replace('₹', '', regex=False) \
                        .str.replace('$', '', regex=False) \
                        .str.replace('rs.', '', regex=False) \
                        .str.replace(',', '', regex=False) \
                        .str.replace('inr', '', regex=False) \
                        .str.replace('(ex-showroom)', '', regex=False) \
                        .str.replace('(expected)', '', regex=False) \
                        .str.strip()

# 2. Handle ranges (take the first value in a range)
df['price'] = df['price'].apply(lambda x: x.split('-')[0].strip() if isinstance(x, str) and '-' in x else x)

# 3. Convert 'lakh' values to numeric and multiply by 100,000
def convert_lakhs_and_to_numeric(price_str):
    if isinstance(price_str, str):
        price_str = price_str.strip()
        # Check if 'lakh' is present
        if 'lakh' in price_str:
            num_part = price_str.replace('lakh', '').strip()
            try:
                return (num_part) * 100000
            except ValueError:
                return price_str
        else:
            try:
                return (price_str)
            except ValueError:
                return price_str
    return price_str

df['price'] = df['price'].apply(convert_lakhs_and_to_numeric)

# 4. Convert to numeric, coercing any remaining errors to NaN
df['price'] = pd.to_numeric(df['price'], errors='coerce')

# 5. Drop rows with NaN prices after conversion, as these are unparseable values
df.dropna(subset=['price'], inplace=True)

# 6. Ensure 'x.xx' format by rounding to 2 decimal places (underlying type is float)
df['price'] = df['price'].round(2)

df['price'].unique()
```
